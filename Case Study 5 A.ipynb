### From the data provided on Hollywood movies:

```python
import numpy
import pandas as pd
hollywood_df = pd.read_csv('./files/HollywoodMovies.csv')
#hollywood_df.info()
hollywood_df
```

#### 1.Find the highest-rated movie in the“Quest” story type.

```python
quest_df = hollywood_df[hollywood_df.Story == 'Quest']
quest_df.sort_values(by='RottenTomatoes', ascending=False).iloc[0]['Movie']
```

#### 2.Find the genre in which there has been the greatest number of movie releases

```python
hollywood_df.groupby("Genre",as_index=False)["Movie"].count().max()['Genre']
```

#### 3.Print the names of the top five movies with the costliest budgets.

```python
from matplotlib import pyplot as plt

bar = hollywood_df.sort_values(by="Budget",ascending=False)[['Movie','Budget']].head(5)[::-1]

fig, ax = plt.subplots(figsize=(17, 5))
hbar = ax.barh(bar.Movie,bar.Budget,color='green')
ax.bar_label(hbar, label_type='edge', padding=5)
ax.set_xlabel('Movie')
ax.set_ylabel('Budget')
ax.set_title('Movie Budget Chart with Data Labels')
ax.set_xlim(right=max(bar['Budget']) + 50)
plt.show()
```

#### 4.Is there any correspondence between the critics’ evaluation of a movie and its acceptance by the public? Find out, by plotting the net profitability of a movie against the ratings it receives on Rotten Tomatoes.

```python
from matplotlib import pyplot as plt
plt.figure(figsize=(17,5))
plt.scatter(hollywood_df["RottenTomatoes"], hollywood_df["AudienceScore"],color="green")
plt.title("Critics evaluation vs acceptance by public")
plt.show()

#Yes there is a relation between them... Critics evaluation and public response is directly proportional to each other.
```

```python
from matplotlib import pyplot as plt
plt.figure(figsize=(17,5))
plt.scatter(hollywood_df["RottenTomatoes"], hollywood_df["Profitability"],color="green")
plt.title("Net Profitability against Rating on Rotten Tomatoes")
plt.ylim(0,2000)
plt.show()
```

### Perform Operations on Files

```python
# From the raw data below create a Pandas Series
import numpy as np
import pandas as pd
series_df = pd.Series(['Amit', 'Bob', 'Kate', 'A', 'b', np.nan, 'Car', 'dog', 'cat'])

# Print all elements in lower case
print(series_df.str.lower())

# Print all the elements in upper case
print(series_df.str.upper())

# Print the length of all the elements
print(series_df.str.len())
```

```python
# From the raw data below create a Pandas Series
import pandas as pd
import numpy as np

series_df_1 = pd.Series([' Atul', 'John ', ' jack ', 'Sam'])

# Print all elements after stripping spaces from the left and right
print(series_df_1.str.strip())

# Print all the elements after removing spaces from the left only
print(series_df_1.str.lstrip())

# Print all the elements after removing spaces from the right only
print(series_df_1.str.rstrip())
```

```python
# Create a series from the raw data below
import numpy as np
import pandas as pd
series_df_2 = pd.Series(['India_is_big', 'Population_is_huge', np.nan, 'Has_diverse_culture'])

# split the individual strings wherever ‘_’ comes and create a list out of it.
print(series_df_2.str.split('_'))

# Access the individual elements of a list
print(series_df_2.str.split("_").str[1])

# Expand the elements so that all individual elements get split by ‘_’ and insteadof list returns individual elements
print(series_df_2.str.split('_').explode())
```

```python
# Create a series and replace either X or dog with XX-XX
import pandas as pd
import numpy as np
series_df_3 = pd.Series(['A', 'B', 'C', 'AabX', 'BacX','', np.nan, 'CABA', 'dog', 'cat'])
series_df_3 = series_df_3.str.replace(r'(X)','XX-XX',regex=True)
series_df_3 = series_df_3.str.replace('dog','XX-XX',case=True)
print(series_df_3)

# Create a series and remove the dollars from the numeric values
series_df_3 = pd.Series(['12', '-$10', '$10,000'])
series_df_3 = series_df_3.str.replace(r'$','')
print(series_df_3)
```

```python
# Create a series and reverse all lower case words
import pandas as pd
import numpy as np
series_df_4 = pd.Series(['India1998', 'big country', np.nan])
print(series_df_4.apply(lambda x:x[::-1] if pd.notna(x) and x.islower() else x))
```

```python
# Create pandas series and print true if the value is alphanumeric in series or false if the value is not alphanumeric in series.
import pandas as pd
series_df_5 = pd.Series(['1', '2_', '1a', '2b', '2003c'])
print(series_df_5.apply(lambda x:True if x.isalnum() else False))
```

```python
# Create pandas series and print true if the value is containing ‘A’
import pandas as pd
import numpy as np
series_df_6 = pd.Series(['1', '2', '1a', '2b', 'America', 'VietnAm','vietnam', '2003c'])
print(series_df_6.str.contains('A'))
```

```python
# Create pandas series and print in three columns value 0 or 1 is a or b or c exists in values
import pandas as pd
import numpy as np

#to remove te warning message
pd.set_option('future.no_silent_downcasting', True)

series_df_7 = pd.Series(['a', 'a|b', np.nan, 'a|c'])
series_df_7 = series_df_7.str.contains(r'a|b|c',regex=True).fillna(False).astype(int)
pd.DataFrame(series_df_7).transpose()
```

```python
# Create pandas dataframe having keys and ltable and rtable as below -
left = {'key': ['One', 'Two'], 'ltable': [1, 2]}
right = {'key': ['One', 'Two'], 'rtable': [4, 5]}

# Merge both the tables based onkey
import pandas as pd
left_df = pd.DataFrame(left)
right_df = pd.DataFrame(right)
print(left_df.merge(right_df, how='inner', on='key'))
```
